# Hello Thymio

Notebook counterpart of `hello_thymio.py` script

```python
import math
from pyenki.buffer import EnkiRemoteFrameBuffer
from hello_thymio import setup

orthographic = False
T = 5
dt= 0.1

world = setup()
w = EnkiRemoteFrameBuffer(world=world, camera_position=(0, 0), camera_altitude=70.0,
                          camera_yaw=0.0, camera_pitch=-math.pi / 2,
                          camera_is_ortho=orthographic)
w
```

```python
await w.run_async(time_step=dt, duration=T, factor=1)
```

```python
from pyenki.video import make_video
```

```python
world = setup()

v = make_video(world,
               time_step=0.033,
               duration=2,
               camera_position=(0, 0), camera_altitude=70.0,
               camera_yaw=0.0, camera_pitch=-math.pi / 2,
               camera_is_ortho=orthographic, width=1280, height=720)
```

```python
v.display_in_notebook(fps=30, width=640)
```
