```python
import pyenki

world = pyenki.World()
thymio = pyenki.Thymio2()
thymio.left_wheel_target_speed = 10
world.add_object(thymio)
epuck = pyenki.EPuck()
epuck.position = (20, 10)
epuck.right_wheel_target_speed = 10
world.add_object(epuck)
```

```python
from pyenki.buffer import EnkiRemoteFrameBuffer

w = EnkiRemoteFrameBuffer(world=world)
w.move_camera(target_position=epuck.position, target_altitude=epuck.height/2, 
              yaw=1, pitch=-0.5, target_distance=40)
w
```

```python
w.snapshot()
```

```python
await w.run_async(time_step=0.033, duration=5, factor=1)
```
